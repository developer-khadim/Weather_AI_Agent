```python
from agents import Agent,Runner,function_tool
import asyncio
import requests

@function_tool
def get_weather(city: str) -> str:
    """
    Get current weather info for a city using WeatherAPI.
    """
    try:
        weather_api_key = os.getenv("WEATHER_API_KEY")
        url = f"http://api.weatherapi.com/v1/current.json?key={weather_key}&q={city}"
        response = requests.get(url, timeout=10)
        data = response.json()

        if 'current' not in data:
            return f"Could not find weather data for '{city}'."

        condition = data['current']['condition']['text']
        temp_c = data['current']['temp_c']
        return f"The weather in {city} is {condition} with temperature {temp_c}°C."

    except Exception as e:
        return f"Failed to fetch weather: {str(e)}"


agent = Agent(
    name= "weather-agent",
    model="gpt-4.1-nano",
    instructions = (
    "You are a smart and reliable weather forecast assistant. "
    "Your job is to provide users with accurate, up-to-date weather information, "
    "including temperature, humidity, wind speed, and conditions like rain or sunshine. "
    "Always respond clearly and helpfully, using friendly and concise language. "
    "If a user asks for the weather in a specific city or location, fetch and return the latest data. "
    "Do not guess or make up any weather information. Your goal is to help users plan their day with confidence."
),
    tools=[get_weather]
)

async def main():
    result = await Runner.run(agent, "What is the weather in sukkur right now?")
    print(result.final_output)
    
    
await main()
```
